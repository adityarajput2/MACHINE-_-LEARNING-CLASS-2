POWER TRANSFORMER 

```python
import numpy as np
from sklearn.preprocessing import PowerTransformer

x = np.array([
    [-1],
    [0],
    [3],
    [5],
    [10],
    [50],
])

pt = PowerTransformer(method = "yeo-johnson")

X_transformed = pt.fit_transform(x)

print("original:")
print(x.ravel())
print("\nTransformed:")
print(np.round(X_transformed.ravel(),3))
print("\nLearned lambda:")
print(pt.lambdas_)
```

```python
import numpy as np
from sklearn.preprocessing import PowerTransformer

x = np.array([
    [1],
    [1],
    [3],
    [5],
    [10],
    [50],
])

pt = PowerTransformer(method = "box-cox")

X_transformed = pt.fit_transform(x)

print("original:")
print(x.ravel())
print("\nTransformed:")
print(np.round(X_transformed.ravel(),3))
print("\nLearned lambda:")
print(pt.lambdas_)
```
