## Creating and Working with Pandas DataFrame

### 1. Import Pandas Library

```python
import pandas as pd
import numpy as np
```

### 2. Method 1: Create DataFrame using a Python Dictionary (Most Common)

```python
# Sample data
data = {
    'Name': ['Aditya', 'Rahul', 'Priya', 'Sneha', 'Aman'],
    'Age': [21, 23, 20, 22, 24],
    'Course': ['Data Science', 'Machine Learning', 'AI', 'Data Science', 'Deep Learning'],
    'Score': [88, 92, 95, 89, 94]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display DataFrame
df
```

### 3. Method 2: Create DataFrame from List of Lists with Column Names

```python
data_list = [
    ['Python', 1991, 'Guido van Rossum'],
    ['Java', 1995, 'James Gosling'],
    ['C++', 1985, 'Bjarne Stroustrup']
]

columns = ['Language', 'Year', 'Founder']

df_languages = pd.DataFrame(data_list, columns=columns)
df_languages
```

### 4. Basic DataFrame Operations

```python
# View first few rows
print("--- Head ---")
print(df.head(3))

# Check information about data types and non-null values
print("\n--- Info ---")
print(df.info())

# Statistical summary of numerical columns
print("\n--- Summary Statistics ---")
print(df.describe())
```
